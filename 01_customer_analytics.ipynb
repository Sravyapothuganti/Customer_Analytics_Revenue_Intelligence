```python
sales_df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
].copy()

# Remove exact duplicate transactions
sales_df = sales_df.drop_duplicates()

print("Original rows:", len(df))
print("Clean sales rows:", len(sales_df))
print("Rows excluded:", len(df) - len(sales_df))

print("\nMissing values:")
print(sales_df.isnull().sum())
```

```python
customer_df = sales_df.dropna(subset=["CustomerID"]).copy()

# Convert CustomerID from decimal-looking values to integers
customer_df["CustomerID"] = customer_df["CustomerID"].astype(int)

print("Sales rows:", len(sales_df))
print("Customer analysis rows:", len(customer_df))
print("Unique customers:", customer_df["CustomerID"].nunique())

print("\nCustomer dataset missing values:")
print(customer_df.isnull().sum())
```

```python
sales_df["Revenue"] = sales_df["Quantity"] * sales_df["UnitPrice"]

total_revenue = sales_df["Revenue"].sum()
total_orders = sales_df["InvoiceNo"].nunique()
total_units = sales_df["Quantity"].sum()

print("Total Revenue: $", round(total_revenue, 2))
print("Total Orders:", total_orders)
print("Total Units Sold:", total_units)
```

```python
snapshot_date = pd.Timestamp("2011-12-10")

rfm = (
    customer_df.groupby("CustomerID")
    .agg(
        Recency=("InvoiceDate",
                 lambda x: (snapshot_date - x.max()).days),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("Revenue", "sum")
    )
)

print("RFM dataset shape:", rfm.shape)

print("\nRFM Summary:")
print(rfm.describe().round(2))

print("\nFirst 10 customers:")
print(rfm.head(10).round(2))
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    q=4,
    labels=[4, 3, 2, 1],
    duplicates="drop"
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=4,
    labels=[1, 2, 3, 4]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    q=4,
    labels=[1, 2, 3, 4]
)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

print(
    rfm[
        ["Recency", "Frequency", "Monetary",
         "R_Score", "F_Score", "M_Score", "RFM_Score"]
    ].head(10)
)
```

```python
def segment_customer(row):
    r = int(row["R_Score"])
    f = int(row["F_Score"])
    m = int(row["M_Score"])

    if r == 4 and f >= 3 and m >= 3:
        return "Champions"

    elif r >= 3 and f >= 3 and m >= 2:
        return "Loyal Customers"

    elif r == 4 and f <= 2:
        return "New / Promising"

    elif r == 3 and f <= 2:
        return "Potential Loyalists"

    elif r <= 2 and f >= 3 and m >= 3:
        return "At Risk"

    elif r == 1 and f <= 2:
        return "Lost / Inactive"

    else:
        return "Needs Attention"


rfm["Segment"] = rfm.apply(segment_customer, axis=1)

segment_counts = (
    rfm["Segment"]
    .value_counts()
)

print("Customer Segments:")
print(segment_counts)

print("\nSegment Percentage:")
print(
    (segment_counts / len(rfm) * 100)
    .round(2)
)
```

```python
segment_summary = (
    rfm.groupby("Segment", observed=True)
    .agg(
        Customers=("Monetary", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean"),
        Total_Revenue=("Monetary", "sum")
    )
)

segment_summary["Revenue_Share_%"] = (
    segment_summary["Total_Revenue"]
    / rfm["Monetary"].sum()
    * 100
)

segment_summary = (
    segment_summary
    .sort_values("Total_Revenue", ascending=False)
)

print("RFM Segment Summary:")
print(segment_summary.round(2))
```

## Key Business Insights

### 1. Revenue Performance
- The business generated approximately **$10.64M in positive sales revenue** across **19,960 orders**.
- More than **5.57M units** were sold, with an **average order value of $533.17**.
- Revenue accelerated significantly during September–November 2011, with **November generating approximately $1.50M**, the highest full month in the dataset.
- December 2011 contains only data through December 9 and should not be compared directly with complete months.

### 2. Customer Revenue Concentration
- **4,338 identifiable customers** generated approximately **$8.89M**, representing **83.51% of positive sales revenue**.
- The remaining **16.49%** of revenue cannot be linked to a CustomerID.
- Champions and Loyal Customers represent **34.16% of identifiable customers but generate 73.92% of customer-attributed revenue**.
- The top 10 customers generate **17.30%** of customer-attributed revenue, showing meaningful customer concentration without dependence on only a few accounts.

### 3. Customer Segmentation
- RFM analysis identified **Champions as the highest-value customer segment**, contributing approximately **54.84% of customer-attributed revenue**.
- Loyal Customers contribute another **19.08%**.
- At-Risk customers contribute approximately **10.82%**, making them an important retention opportunity.
- Lost/Inactive and Needs Attention customers represent substantial customer counts but contribute relatively small shares of revenue.

### 4. Geographic Performance
- The **United Kingdom contributes approximately 84.59% of positive sales revenue**, demonstrating strong geographic concentration.
- The Netherlands, EIRE, Germany, France, and Australia are among the strongest international markets.
- The Netherlands and Australia combine relatively high revenue with high average order values, making them potentially valuable international markets.

### 5. Product Performance
- REGENCY CAKESTAND 3 TIER is one of the strongest merchandise products by revenue.
- Several products show strong revenue and order frequency, including WHITE HANGING HEART T-LIGHT HOLDER, PARTY BUNTING, and JUMBO BAG RED RETROSPOT.
- Bulk transactions can significantly distort product rankings, so revenue, units, and order frequency should be evaluated together.

### 6. Cancellations and Anomalies
- Recorded cancellation value totals approximately **$896.8K**.
- Using positive sales plus cancellation value as the comparison base, cancellations represent approximately **7.77% of transaction value**.
- December 2011 initially appears to have an unusually large cancellation spike of **$205.1K**.
- Further investigation showed that a single PAPER CRAFT, LITTLE BIRDIE transaction contributed **82.13% of December's cancellation value**.
- This demonstrates that the December spike was primarily driven by a transaction-level outlier rather than a broad increase in cancellation activity.

## Business Recommendations

1. **Prioritize High-Value Customer Retention**
   - Champions and Loyal Customers generate 73.92% of customer-attributed revenue.
   - Use loyalty incentives, personalized offers, and early-access campaigns to retain these high-value customers.

2. **Re-engage At-Risk Customers**
   - At-Risk customers still contribute 10.82% of customer-attributed revenue.
   - Target these customers with reactivation campaigns based on previous purchasing behavior.

3. **Reduce Geographic Revenue Concentration**
   - The United Kingdom contributes 84.59% of positive sales revenue.
   - Evaluate expansion opportunities in established international markets such as the Netherlands, Germany, France, and Australia.

4. **Monitor Product Performance Using Multiple Metrics**
   - Evaluate products using revenue, units sold, order frequency, and cancellation behavior rather than revenue alone.
   - Separate bulk one-time transactions from recurring product demand when evaluating product performance.

   - Large individual transactions can significantly distort monthly cancellation trends.
   - Implement transaction-level monitoring for unusually large orders, cancellations, and reversals before interpreting them as broader business trends.

6. **Improve Customer Identification**
   - Approximately 16.49% of positive sales revenue cannot be associated with a CustomerID.
   - Improving customer identification would strengthen segmentation, retention analysis, and customer lifetime value measurement.

```python
growth_query = """
WITH monthly_revenue AS (
    SELECT
        YearMonth,
        SUM(Revenue) AS revenue
    FROM sales
    GROUP BY YearMonth
),

revenue_growth AS (
    SELECT
        YearMonth,
        revenue,
        LAG(revenue) OVER (
            ORDER BY YearMonth
        ) AS previous_month_revenue
    FROM monthly_revenue
)

SELECT
    YearMonth,
    ROUND(revenue, 2) AS revenue,
    ROUND(previous_month_revenue, 2) AS previous_month_revenue,
    ROUND(
        ((revenue - previous_month_revenue)
        / previous_month_revenue) * 100,
        2
    ) AS mom_growth_percent
FROM revenue_growth
ORDER BY YearMonth;
"""

growth_sql_results = pd.read_sql_query(
    growth_query,
    conn
)

growth_sql_results
```

```python
customer_share_query = """
WITH customer_revenue AS (
    SELECT
        CustomerID,
        SUM(Revenue) AS revenue
    FROM sales
    WHERE CustomerID IS NOT NULL
    GROUP BY CustomerID
),

ranked_customers AS (
    SELECT
        CustomerID,
        revenue,
        ROW_NUMBER() OVER (
            ORDER BY revenue DESC
        ) AS revenue_rank
    FROM customer_revenue
)

SELECT
    ROUND(
        SUM(
            CASE
                WHEN revenue_rank <= 10 THEN revenue
                ELSE 0
            END
        ),
        2
    ) AS top_10_customer_revenue,

    ROUND(SUM(revenue), 2) AS total_customer_revenue,

    ROUND(
        SUM(
            CASE
                WHEN revenue_rank <= 10 THEN revenue
                ELSE 0
            END
        ) / SUM(revenue) * 100,
        2
    ) AS top_10_revenue_share_percent

FROM ranked_customers;
"""

customer_share_sql = pd.read_sql_query(
    customer_share_query,
    conn
)

customer_share_sql
```
